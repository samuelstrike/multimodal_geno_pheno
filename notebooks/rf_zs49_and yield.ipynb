### Applying Random Forest on gene data only

```python
import numpy as np
import optuna
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

n_splits = 3
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)

def objective(trial):
    n_estimators = trial.suggest_int('n_estimators', 0, 700)
    max_depth = trial.suggest_int('max_depth', 10, 100)
    min_samples_split = trial.suggest_int('min_samples_split', 2, 10)
    min_samples_leaf = trial.suggest_int('min_samples_leaf', 1, 10)

    mse = []

    for train_index, test_index in kf.split(X_yield):
        X_train, X_test = X_yield.iloc[train_index], X_yield.iloc[test_index]
        y_train, y_test = y_yield.iloc[train_index], y_yield.iloc[test_index]

        rf_model_yield = RandomForestRegressor(
            n_estimators=n_estimators,
            max_depth=max_depth,
            min_samples_split=min_samples_split, 
            min_samples_leaf=min_samples_leaf,
            n_jobs=-1,
            random_state=42
            )
        
        rf_model_yield.fit(X_train, y_train)

        y_pred = rf_model_yield.predict(X_test)

        mse.append(mean_squared_error(y_test, y_pred))

    return np.mean(mse)
```

```python
study = optuna.create_study(direction='minimize', study_name='RandomForestRegressorGSyield', storage='sqlite:///optuna_study/RandomForestRegressorGSyield.sqlite3', load_if_exists=True)
study.optimize(objective, n_trials=100, show_progress_bar=True)
```

```python
# use the best hyperparameters to train the model   

best_params = study.best_params
best_params

rf_model = RandomForestRegressor(
    # n_estimators=best_params['n_estimators'],
    # max_depth=best_params['max_depth'],
    # min_samples_split=best_params['min_samples_split'],
    # min_samples_leaf=best_params['min_samples_leaf'],
    n_estimators= 604,
    max_depth=10,
    min_samples_split=2,
    min_samples_leaf=6,
    n_jobs=-1,
    random_state=42
)

train_mse, train_mae, train_r2 = [], [], []
test_mse, test_mae, test_r2 = [], [], []

for train_index, test_index in kf.split(X_yield):
    X_train, X_test = X_yield.iloc[train_index], X_yield.iloc[test_index]
    y_train, y_test = y_yield.iloc[train_index], y_yield.iloc[test_index]

    rf_model.fit(X_train, y_train)

    y_pred_test = rf_model.predict(X_test)
    y_pred_train = rf_model.predict(X_train)

    test_mse.append(mean_squared_error(y_test, y_pred_test))
    test_mae.append(mean_absolute_error(y_test, y_pred_test))
    test_r2.append(r2_score(y_test, y_pred_test))

    train_mse.append(mean_squared_error(y_train, y_pred_train))
    train_mae.append(mean_absolute_error(y_train, y_pred_train))
    train_r2.append(r2_score(y_train, y_pred_train))
```

```python
print(f'Train MSE: {np.mean(train_mse):.2f}')
print(f'Train RMSE: {np.sqrt(np.mean(train_mse)):.2f}')
print(f'Train MAE: {np.mean(train_mae):.2f}')
print(f'Train R2: {np.mean(train_r2):.2f}\n')
print(f'Standard deviation of Train MSE: {np.std(train_mse):.2f}\n')

print(f'Test MSE: {np.mean(test_mse):.2f}')
print(f'Test RMSE: {np.sqrt(np.mean(test_mse)):.2f}')
print(f'Test MAE: {np.mean(test_mae):.2f}')
print(f'Test R2: {np.mean(test_r2):.2f}')
print(f'Standard deviation of Test MSE: {np.std(test_mse):.2f}')

```
