### Fusion Model -LASSO on LSTM and RF

```python
import numpy as np
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam, SGD
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.layers import LSTM, Dense, Input, Concatenate, Dropout, BatchNormalization, Normalization, Embedding, Flatten
from sklearn.model_selection import KFold
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import tensorflow as tf
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import Lasso

n_splits = 3
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)
    
val_losses = []

mse_train_ensemble, mae_train_ensemble, r2_train_ensemble = [], [], []
mse_test_ensemble, mae_test_ensemble, r2_test_ensemble = [], [], []

for train_index, test_index in kf.split(concatenated_data):
    X_train_lstm, X_test_lstm = concatenated_data[train_index], concatenated_data[test_index]
    X_train_rf, X_test_rf = X_yield.iloc[train_index], X_yield.iloc[test_index]
    X_study_train, X_study_test = study[train_index], study[test_index]
    y_train, y_test = target[train_index], target[test_index]

    print(X_train_lstm.shape, X_test_lstm.shape, X_train_rf.shape, X_test_rf.shape, X_study_train.shape, X_study_test.shape, y_train.shape, y_test.shape)
        
    # Scale data
    scaler = MinMaxScaler()
    X_train_scaled_lstm = scaler.fit_transform(X_train_lstm.reshape(-1, X_train_lstm.shape[-1])).reshape(X_train_lstm.shape)
    X_test_scaled_lstm = scaler.transform(X_test_lstm.reshape(-1, X_test_lstm.shape[-1])).reshape(X_test_lstm.shape)
    # X_study_train_scaled = scaler.fit_transform(X_study_train.reshape(-1, 1))
    # X_study_test_scaled = scaler.transform(X_study_test.reshape(-1, 1))
    
    input_shape = (4, 3)  # (time_steps, features)
    lstm_input = Input(shape=input_shape)
    study_input = Input(shape=(1,))
    study_embedded = Embedding(input_dim=6, output_dim=3, input_length=1)(study_input)
    study_embedded = Flatten()(study_embedded)

    lstm_out = LSTM(223, return_sequences=True)(lstm_input)
    dropout1 = Dropout(0.2)(lstm_out)
    lstm_out = LSTM(265, return_sequences=False)(dropout1)
    dropout2 = Dropout(0.2)(lstm_out)

    concat = Concatenate()([dropout2, study_embedded])

    #scale the concatenated data
    normalizer = Normalization()
    normalized_concat = normalizer(concat)

    dense_out = Dense(136, activation='relu')(normalized_concat)
    dropout3 = Dropout(0.3)(dense_out)
    final_output = Dense(1, activation='relu')(dropout3)

    model = Model(inputs=[lstm_input, study_input], outputs=final_output)

    # optimizer = SGD(learning_rate=0.0009, momentum=0.77)

    optimizer = Adam(learning_rate=0.001)

    model.compile(optimizer=optimizer, loss='mse', metrics=[tf.keras.metrics.RootMeanSquaredError()])
    
    early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    
    # Train model
    history = model.fit(
        [X_train_scaled_lstm, X_study_train],
        y_train,
        batch_size=32,
        epochs=100,
        validation_split=0.2,
        # callbacks=[early_stopping],
        verbose=1
    )

    lr = RandomForestRegressor(
        n_estimators=604,
        max_depth=10,
        min_samples_split=2, 
        min_samples_leaf=9,
        n_jobs=-1,
        random_state=42
    )
    
    lr.fit(X_train_rf, y_train)

    train_preds_rf = lr.predict(X_train_rf)
    test_preds_rf = lr.predict(X_test_rf)

    # Evaluate model
    # val_loss = model.evaluate([X_test_scaled, X_study_test], y_test)[0]
    # val_losses.append(val_loss)

    train_preds_lstm = model.predict([X_train_scaled_lstm, X_study_train])
    test_preds_lstm = model.predict([X_test_scaled_lstm, X_study_test])
    # Combine predictions from LSTM and RandomForest
    combined_preds_train =np.column_stack((train_preds_lstm, train_preds_rf))
    combined_preds_test = np.column_stack((test_preds_lstm, test_preds_rf))

    # Perform Lasso regression
    lasso = Lasso(alpha=0.003)
    lasso.fit(combined_preds_train, y_train)

    train_preds_ensemble = lasso.predict(combined_preds_train)
    test_preds_ensemble = lasso.predict(combined_preds_test)

    mse_train_ensemble.append(mean_squared_error(y_train, train_preds_ensemble))
    mae_train_ensemble.append(mean_absolute_error(y_train, train_preds_ensemble))
    r2_train_ensemble.append(r2_score(y_train, train_preds_ensemble))

    mse_test_ensemble.append(mean_squared_error(y_test, test_preds_ensemble))
    mae_test_ensemble.append(mean_absolute_error(y_test, test_preds_ensemble))
    r2_test_ensemble.append(r2_score(y_test, test_preds_ensemble))


   
```

```python
print(f'Train MSE: {np.mean(mse_train_ensemble):.2f}')
print(f'Train RMSE: {np.sqrt(np.mean(mse_train_ensemble)):.2f}')
print(f'Train MAE: {np.mean(mae_train_ensemble):.2f}')
print(f'Train R2: {np.mean(r2_train_ensemble):.2f}\n')
print(f'Standard deviation of Train MSE: {np.std(mse_train_ensemble):.2f}\n')

print(f'Test MSE: {np.mean(mse_test_ensemble):.2f}')
print(f'Test RMSE: {np.sqrt(np.mean(mse_test_ensemble)):.2f}')
print(f'Test MAE: {np.mean(mae_test_ensemble):.2f}')
print(f'Test R2: {np.mean(r2_test_ensemble):.2f}')
print(f'Standard deviation of Test MSE: {np.std(mse_test_ensemble):.2f}')

```

### Intermediate fusion using LSTM and CNN

```python
from tensorflow.keras.layers import LSTM, Dropout, Dense, Input, concatenate, BatchNormalization, Conv1D, Activation, MaxPooling1D, Flatten
from tensorflow.keras.optimizers import Adam, SGD
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import EarlyStopping
import tensorflow as tf

def intermediate_fusion(input_lstm, input_cnn, input_study):
    """
    This function should implement the intermediate fusion of the two inputs.
    """
    lstm_input = Input(shape = input_lstm)
    cnn_input = Input(shape = input_cnn)
    study_input = Input(shape = input_study)

    study_embedded = Embedding(input_dim=6, output_dim=3, input_length=1)(study_input)
    study_embedded = Flatten()(study_embedded)

    # Implement the intermediate fusion of the two inputs
    x = LSTM(223, return_sequences=True)(lstm_input)
    x = Dropout(0.2)(x)
    x = LSTM(265, return_sequences = False)(x)
    x = Dropout(0.2)(x)
    contact_study = concatenate([x,study_embedded])
    contact_study = BatchNormalization()(contact_study)
    x1 = Dense(136, activation='relu')(contact_study)
    x1 = Dropout(0.3)(x1)
    lstm_output = Dense(128, activation='relu')(x1)

    y = Conv1D(filters=46, kernel_size=3, strides =1, padding ='same')(cnn_input)
    y = BatchNormalization()(y)
    y = Activation('relu')(y)
    y = MaxPooling1D(pool_size=2)(y)
    y = Flatten()(y)
    y = Dropout(0.4)(y)
    y = Dense(464)(y)
    y = Activation('relu')(y)
    cnn_output = Dense(128)(y)

    concatenated = concatenate([lstm_output, cnn_output])

    normalizer = Normalization()
    normalized_concat = normalizer(concatenated)

    z = Dense(128, activation='relu')(normalized_concat)
    z = Dropout(0.2)(z)
    final_output = Dense(1)(z)

    intermediate_fusion = Model(inputs=[lstm_input, cnn_input, study_input], outputs=final_output)

    optimizer = Adam(learning_rate=0.001)
    # optimizer = SGD(learning_rate=0.0009, momentum=0.77)

    intermediate_fusion.compile(optimizer=optimizer, loss='mean_squared_error', metrics=[tf.keras.metrics.RootMeanSquaredError()])    
    
    return intermediate_fusion
```

```python
from sklearn.model_selection import KFold

kf = KFold(n_splits=3, shuffle=True, random_state=42)

final_train_mse, final_train_mae, final_train_r2 = [], [], []
final_test_mse,  final_test_mae, final_test_r2 = [], [],[]

for train_index, test_index in kf.split(concatenated_data):
    X_train_lstm, X_test_lstm = concatenated_data[train_index], concatenated_data[test_index]
    X_train_cnn, X_test_cnn = X_zs49_cnn[train_index], X_zs49_cnn[test_index]
    X_study_train, X_study_test = study[train_index], study[test_index]
    y_train, y_test = target[train_index], target[test_index]

    # Scale data
    scaler = MinMaxScaler()
    X_train_scaled_lstm = scaler.fit_transform(X_train_lstm.reshape(-1, X_train_lstm.shape[-1])).reshape(X_train_lstm.shape)
    X_test_scaled_lstm = scaler.transform(X_test_lstm.reshape(-1, X_test_lstm.shape[-1])).reshape(X_test_lstm.shape)

    input_shape_lstm = (5, 3)  # (time_steps, features)
    input_shape_cnn = (X_train_cnn.shape[1], 1)  # (time_steps, features)
    input_shape_study = (1,)

    intermediate_fusion_model = intermediate_fusion(input_shape_lstm, input_shape_cnn, input_shape_study)

    early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)

    history = intermediate_fusion_model.fit(
        [X_train_scaled_lstm, X_train_cnn, X_study_train],
        y_train,
        batch_size=32,
        epochs=100,
        validation_split=0.2,
        callbacks=[early_stopping],
        verbose=1
    )

    final_y_pred_test = intermediate_fusion_model.predict([X_test_scaled_lstm, X_test_cnn, X_study_test])
    final_y_pred_train = intermediate_fusion_model.predict([X_train_scaled_lstm, X_train_cnn, X_study_train])

    final_train_mse.append(mean_squared_error(y_train, final_y_pred_train))
    final_train_mae.append(mean_absolute_error(y_train, final_y_pred_train))
    final_train_r2.append(r2_score(y_train, final_y_pred_train))

    final_test_mse.append(mean_squared_error(y_test, final_y_pred_test))
    final_test_mae.append(mean_absolute_error(y_test, final_y_pred_test))
    final_test_r2.append(r2_score(y_test, final_y_pred_test))

```
